# Implementações Técnicas e Análise Qualitativa:

Implementações Técnicas que a gente usou na prática:

- Ajuste de pesos (weights=[0.5, 0.5]): Usado lá na criação dos dados para deixar a base bem equilibrada entre as classes (sem viés de amostragem).

- Método lr.coef_ (Regressão Logística): Serve para puxar os pesos e coeficientes diretos, provando o quanto o modelo é enxuto e compacto.

- Método nb.class_prior_ (Naive Bayes): Pra ver as probabilidades iniciais a priori que o modelo calcula.

- Método predict_proba (y_proba_nb): Usado para extrair as probabilidades e analisar o nível de confiança que o modelo tem nas predições de fraude.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.decomposition import PCA

# Gerar dados sintéticos simulando transações financeiras
X, y = make_classification(n_samples=2000, n_features=20, n_informative=10, n_redundant=5,
                           n_clusters_per_class=2, weights=[0.5, 0.5], flip_y=0.01, random_state=42)

# Reduzir dimensão para visualização
X_vis = PCA(n_components=2).fit_transform(X)

# Dividir em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

X_test[0] += 10

# Treinar Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred_nb = nb.predict(X_test)
y_proba_nb = nb.predict_proba(X_test)[:, 1]

# Treinar Regressão Logística
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
y_proba_lr = lr.predict_proba(X_test)[:, 1]

# Métricas e inferência
print("Regressão Logística\n", classification_report(y_test, y_pred_lr))
print("Naive Bayes\n", classification_report(y_test, y_pred_nb))

# Para ver a quantidade de pesos da Regressão Logística
print("Quantidade de pesos da regressão logística", lr.coef_)

# Para ver a probabilidade inicial das classes no Naive Bayes
print("probabilidade inicial das classes no Naive Bayes", nb.class_prior_)

# metodos para verificar se é realmente fraudulenta
print("probabilidade de fraude do Naive Bayes", y_proba_nb)
```

## 1. O conjunto de dados é desbalanceado (90% não fraudulento). Como isso pode impactar os modelos? Qual modelo parece lidar melhor com o desbalanceamento?

R= O desbalanceamento deixa o modelo "preguiçoso", porque como 90% dos dados (0.9) são normais, ele aprende que chutar sempre "não fraude" (0 no caso) já dá uma acurácia alta, mesmo errando as fraudes de verdade. Isso estraga as métricas da classe menor. Olhando os relatórios, a Regressão Logística lidou melhor com isso, principalmente porque ela teve uma precisão bem maior na classe de fraude (0.75 contra 0.49 do Naive Bayes), ou seja, ela deu menos alarmes falsos.

## 2. Altere o parâmetro weights para [0.5, 0.5] e reexecute. O desempenho melhora? O comportamento dos classificadores muda?

R= Bom, primeiramente é importante a compreensão do que foi feito. Anteriormente nossos pesos (WEIGHTS) eram 0.9 = 90% e 0.1 = 10%, conforme foi adaptado seguindo o que a questão pede ficou 50% pra cada, o que tirou o desbalanceamento e deixou o famoso "metade/metade". O desempenho melhorou muito, inclusive a classe minoritária no geral agora nos passa mais confiança, pois o modelo viu os dois mundos (não fraude/fraude), métricas baixas agora já estão mais balanceadas em comparação ao que era anteriormente. Sobre os classificadores, mudou sim: ele passou a compreender exatamente o que é uma fraude e o que não é, garantindo uma melhor consistência para os modelos.

## 3. Compare o número de parâmetros dos dois modelos (use .coef_ e .class_prior_). Qual modelo parece mais "compacto"? O que isso implica em termos de uso prático?

R= Olhando os resultados das funções do scikit-learn, a Regressão Logística acaba sendo mais compacta. Visualmente, ela já joga todos os seus pesos direto na tela (.coef_), enquanto o Naive Bayes esconde mais coisas. Mesmo parecendo menor na chamada do .class_prior_, o Naive Bayes carrega uma carga estatística gigante por trás (como médias e variâncias de cada atributo para cada classe). Na prática, a Regressão Logística ganha porque é mais enxuta, gasta menos memória e deixa o sistema muito mais leve e rápido para rodar.

## 4. Altere algumas amostras em X_test para valores extremos (ex: X_test[0] += 10). Como cada modelo reage? Qual é mais robusto a outliers?

R= Após alterar os valores extremos (X_test[0] += 10), dá para ver claramente que o Naive Bayes sentiu bastante o impacto, com uma queda nas métricas, porque ele é mais sensível a valores discrepantes. Já a Regressão Logística praticamente manteve os seus números intactos. Sendo assim, o modelo mais robusto a outliers é a Regressão Logística, já que ela aguenta a presença de dados extremos sem perder a estabilidade, enquanto o Naive Bayes acaba sofrendo mais com a alteração.

## 5. Use os métodos predict_proba para interpretar a confiança de cada modelo nas predições. As probabilidades geradas parecem calibradas (por exemplo, uma transação com 0.9 de chance de fraude é mesmo fraudulenta)?

R= Analisando os valores gerados pelo predict_proba, percebemos que o Naive Bayes tende a ser extremamente confiante (ou polarizado), gerando muitas probabilidades muito próximas de 0 ou de 1 (como 0.99 ou valores bem próximos de zero). Na prática, essas probabilidades nem sempre são perfeitamente calibradas: uma transação com 0,9 de chance de fraude expressa a forte certeza estatística do modelo com base nas distribuições de média e variância, mas isso não significa uma garantia absoluta de 100%, havendo ainda margem para falsos positivos em um cenário real
